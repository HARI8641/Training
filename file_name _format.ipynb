```python
from pathlib import Path

def verified_file_path(allowed_extensions: tuple = None) -> Path:
    """
    Prompts the user for a file path, strips accidental quotes,
    and verifies that the file exists and has the correct format.
    """
    while True:
        raw_path = input("Enter file path: ").strip().strip('"\'')
        path_obj = Path(raw_path)

        if not raw_path:
            print("Path cannot be empty. Please try again.\n")
            continue

        if not path_obj.exists():
            print(f"File not found: '{path_obj}'. Please check the path.\n")
            continue

        if not path_obj.is_file():
            print(f"'{path_obj}' is a directory, not a file. Please enter a valid file path.\n")
            continue

        # Optional: Verify specific file extension (e.g., ('.csv', '.json', '.txt'))
        if allowed_extensions and path_obj.suffix.lower() not in allowed_extensions:
            print(f"Invalid format. Allowed formats: {allowed_extensions}\n")
            continue

        return path_obj.resolve()


def raw_data():
    file_path = verified_file_path(allowed_extensions=('.csv', '.txt', '.json'))

    with open(file_path, "r", encoding="utf-8") as f:
        data = f.read()

    return data, file_path
```

```python
from pathlib import Path

def choose_format(allowed_extensions: tuple) -> str:
    """Displays a numbered menu (CLI dropdown) to select an extension."""
    print("Choose a file format:")
    for idx, ext in enumerate(allowed_extensions, 1):
        print(f"  [{idx}] {ext}")

    while True:
        choice = input(f"Select an option (1-{len(allowed_extensions)}): ").strip()
        if choice.isdigit() and 1 <= int(choice) <= len(allowed_extensions):
            return allowed_extensions[int(choice) - 1]
        print("Invalid choice. Please enter a valid number.\n")


def get_verified_save_path(allowed_extensions: tuple = ('.txt', '.csv', '.json'), fixed_format: str = None) -> Path:
    """
    Handles format selection (fixed or menu-driven) and validates
    the save directory, file extension, and overwrite permissions.
    """
    # 1. Determine the target format
    if fixed_format:
        chosen_ext = fixed_format if fixed_format.startswith('.') else f".{fixed_format}"
    elif allowed_extensions:
        chosen_ext = choose_format(allowed_extensions)
    else:
        chosen_ext = ""

    # 2. Get and validate destination path
    while True:
        raw_path = input(f"\nEnter destination filename/path (will save as {chosen_ext}): ").strip().strip('"\'')

        if not raw_path:
            print("Path cannot be empty. Please try again.")
            continue

        path_obj = Path(raw_path)

        # Force the chosen extension if missing or different
        if chosen_ext and path_obj.suffix.lower() != chosen_ext.lower():
            path_obj = path_obj.with_suffix(chosen_ext)

        # Ensure parent folder exists (or create it)
        parent_dir = path_obj.parent
        if not parent_dir.exists():
            create_dir = input(f"Directory '{parent_dir}' does not exist. Create it? (y/n): ").strip().lower()
            if create_dir == 'y':
                parent_dir.mkdir(parents=True, exist_ok=True)
            else:
                print("Save aborted. Try entering a different folder path.\n")
                continue

        # Prevent accidental file overwrites
        if path_obj.exists():
            overwrite = input(f"'{path_obj.name}' already exists. Overwrite? (y/n): ").strip().lower()
            if overwrite != 'y':
                print("Choose a different filename.\n")
                continue

        return path_obj.resolve()


def save_data(data: str, fixed_format: str = None):
    """Saves data to a verified user path."""
    save_path = get_verified_save_path(
        allowed_extensions=('.txt', '.csv', '.json'),
        fixed_format=fixed_format
    )

    with open(save_path, "w", encoding="utf-8") as f:
        f.write(data)

    print(f"Data successfully saved to: {save_path}")
    return save_path
```
