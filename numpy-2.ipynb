```python


# -------------------------------------------------------------------------
# 1. CREATE TWO 2D ARRAYS (The Grids)
# -------------------------------------------------------------------------
# Let's make a simplified 3x3 Sudoku game (0 represents an empty space)
sudoku_game = np.array([
    [5, 3, 0],
    [6, 0, 0],
    [0, 9, 8]
])

# The matching 3x3 solution grid
sudoku_solution = np.array([
    [5, 3, 4],
    [6, 7, 2],
    [1, 9, 8]
])

print("--- Step 1: Our Individual 2D Arrays ---")
print("Game Grid Shape:", sudoku_game.shape)
print("Solution Grid Shape:", sudoku_solution.shape)
print("\n")

# -------------------------------------------------------------------------
# 2. STACK THEM INTO A 3D ARRAY (The Stack)
# -------------------------------------------------------------------------
# np.stack creates a new axis. By default (axis=0), it puts them as layers.
game_and_solution = np.stack([sudoku_game, sudoku_solution])

print("--- Step 2: The Combined 3D Array ---")
print("3D Array Shape:", game_and_solution.shape) 
# Shape output will be (2, 3, 3) -> 2 layers, 3 rows, 3 columns
print(game_and_solution)
print("\n")

# -------------------------------------------------------------------------
# 3. PRACTICE ACCESSING THE LAYERS (The Use Cases)
# -------------------------------------------------------------------------
print("--- Step 3: Accessing Specific Layers ---")

# Access Layer 0 (The Game)
print("Layer 0 (The Game Grid):")
print(game_and_solution[0])
print("\n")

# Access Layer 1 (The Solution)
print("Layer 1 (The Solution Grid):")
print(game_and_solution[1])
print("\n")

# -------------------------------------------------------------------------
# 4. DEEPER DRILL: Accessing a single number inside the 3D structure
# -------------------------------------------------------------------------
print("--- Step 4: Targeting a single value [layer, row, col] ---")

# Let's check the very top-left number of the puzzle (Row 0, Col 0)
game_starting_number = game_and_solution[0, 0, 0]
solution_correct_number = game_and_solution[1, 0, 0]

print(f"Top-left value in game: {game_starting_number}")
print(f"Top-left value in solution: {solution_correct_number}")

# Let's check the empty space in the middle of the game (Row 1, Col 1)
# and see what the answer is supposed to be.
center_puzzle_space = game_and_solution[0, 1, 1]
center_solution_answer = game_and_solution[1, 1, 1]

print(f"Center space in puzzle: {center_puzzle_space} (Empty)")
print(f"Center space answer: {center_solution_answer} (Correct Answer!)")
```
